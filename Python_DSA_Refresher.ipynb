```python
# Question 1
# This program calculates the number of seconds in a given number of days.

# Ask the user to enter the number of days
days = int(input("Enter the number of days: "))

# Calculate the number of seconds
seconds = days * 24 * 60 * 60

# Display the result
print("The number of seconds is:", seconds)
```

```python
# Question 2
# This program calculates the volume of a sphere.

# Import the math module to use the value of pi
import math

# Ask the user to enter the radius
radius = float(input("Enter the radius of the sphere: "))

# Calculate the volume using r**3
volume = (4 / 3) * math.pi * (radius ** 3)

# Display the result
print("The volume of the sphere is:", volume)
```

```python
# Question 3
# This program calculates the area and perimeter of a square using functions.

# Function to calculate the area of a square
def calculate_area(side):
    return side ** 2

# Function to calculate the perimeter of a square
def calculate_perimeter(side):
    return 4 * side

# Ask the user to enter the side length
side = float(input("Enter the side length of the square: "))

# Call the functions
area = calculate_area(side)
perimeter = calculate_perimeter(side)

# Display the results
print("Area of the square:", area)
print("Perimeter of the square:", perimeter)
```

```python
# Question 4
# This program determines whether a character is uppercase or lowercase.

# Function to check the character
def check_character(character):

    if character.isupper():
        return "The character is uppercase."

    elif character.islower():
        return "The character is lowercase."

    else:
        return "The input is not a letter."

# Ask the user to enter a character
character = input("Enter a character: ")

# Call the function and display the result
result = check_character(character)

print(result)
```

```python
# Question 5
# This program is the Python equivalent of the given pseudocode.

# Set the initial values
x = 0
y = 20

# Repeat the instructions until y is less than 6
while True:

    # Subtract 4 from y
    y = y - 4

    # Add 2/y to x
    x = x + (2 / y)

    # Stop when y is less than 6
    if y < 6:
        break

# Display the value of x
print("The value of x is:", x)
```

```python
# Question 6
# This program accepts 5 values, stores them in a list,
# and calculates their average.

# Create an empty list
values = []

# Use a loop to get 5 values from the user
for i in range(5):

    # Ask the user to enter a value
    value = float(input("Enter value " + str(i + 1) + ": "))

    # Add the value to the list
    values.append(value)

# Calculate the sum of the values
total = sum(values)

# Calculate the average
average = total / len(values)

# Display the values and average
print("Values entered:", values)
print("Average:", average)
```
